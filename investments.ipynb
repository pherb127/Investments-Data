## Step 1 : Load the data 

```python
import wrds
import pandas as pd
from datetime import datetime

# Connect to WRDS
db = wrds.Connection(wrds_username='pherb127')
```

```python
# Import Country Data
data=db.raw_sql("SELECT * FROM wrdsapps_windices.mwcountryreturns "
               "WHERE Country IN ('AUSTRALIA','FRANCE','GERMANY','JAPAN','SWITZERLAND','UNITED KINGDOM') "
               "AND Date BETWEEN '2002-01-01' AND '2024-12-31'")
data.to_csv('country_data.csv', sep=';', index=False)
data = pd.read_csv("country_data.csv", sep=';') 
```

```python
# Import US Data
# 2. Download the monthly value-weighted total return
crsp_vw = db.raw_sql(
    """
    SELECT
      date,
      vwretd      -- value-weighted total return (with dividends)
    FROM crsp.msi
    WHERE date BETWEEN '2002-01-01' AND '2024-12-31'
    """,
    date_cols=['date']
)

# 3. Save to CSV
crsp_vw.to_csv('crsp_value_weighted_returns.csv', sep=";", index=False)
crsp = pd.read_csv("crsp_value_weighted_returns.csv", sep=';') 
```
